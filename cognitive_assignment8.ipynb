## Q1. Text preprocessing
Write an original paragraph of 5–6 sentences. Convert it to lowercase, remove punctuation, tokenize it, compare split() and word_tokenize(), remove stopwords, and show word frequencies.

```python
# Q1(1)
ROLL_NO = 1024170395
NAME = "Paras"

import re
import nltk

for package in ["punkt_tab", "stopwords", "wordnet"]:
    nltk.download(package, quiet=True)

paragraph = (
    "Python is my favorite programming language because it makes small ideas easy to test. "
    "I enjoy building useful tools that solve problems in my daily routine. "
    "Last weekend, I created a simple study planner for my classes. "
    "The program saves tasks, checks dates, and reminds me to take short breaks. "
    "It isn't perfect, but each mistake teaches me something new. "
    "Writing clear code helps me think carefully and share my work with friends."
)

lowercase_text = paragraph.lower()
clean_text = re.sub(r"[^a-z\s]", "", lowercase_text)

print("Name:", NAME)
print("Roll number:", ROLL_NO)
print("Original paragraph:", paragraph)
print("Lowercase text:", lowercase_text)
print("Text without punctuation:", clean_text)
```

```python
# Q1(2)
from nltk.tokenize import word_tokenize, sent_tokenize

sentences = sent_tokenize(paragraph)
words = word_tokenize(clean_text)

print("Sentences:", sentences)
print("Number of sentences:", len(sentences))
print("Words from the cleaned text:", words)
print("Sentences use the original paragraph because it still has sentence punctuation.")
```

```python
# Q1(3)
split_words = paragraph.split()
nltk_words = word_tokenize(paragraph)

print("split():", split_words)
print("word_tokenize():", nltk_words)
print("split() separates text at whitespace and keeps punctuation attached.")
print("word_tokenize() separates punctuation and splits isn't into is and n't.")
print("For example, split() keeps weekend, together, while NLTK separates weekend and ,.")
```

```python
# Q1(4)
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))
filtered_words = []

for word in words:
    if word not in stop_words:
        filtered_words.append(word)

print("Words without stopwords:", filtered_words)
```

```python
# Q1(5)
from nltk import FreqDist
import pandas as pd
from IPython.display import display

frequency = FreqDist(filtered_words)
frequency_table = pd.DataFrame(frequency.most_common(), columns=["Word", "Frequency"])
display(frequency_table)
```

## Q2. Stemming and lemmatization
Use the same paragraph. Extract alphabetic words, remove stopwords, apply Porter stemming and WordNet lemmatization, and compare the results.

## Q3. Bag of Words and TF-IDF
Choose three short texts. Build Bag of Words and TF-IDF representations, then print and interpret the top three TF-IDF keywords from each text.

```python
# Q3(1)
from sklearn.feature_extraction.text import CountVectorizer

texts = [
    "The camera takes sharp photos. The camera lens makes photos bright, and the lens is clear.",
    "The laptop runs fast. The laptop processor makes coding smooth, and coding uses the processor well.",
    "The headphones give clear sound. The headphones cushions feel soft, and the cushions improve sound."
]
text_names = ["Text 1", "Text 2", "Text 3"]

for name, text in zip(text_names, texts):
    print(name + ":", text)

count_vectorizer = CountVectorizer(stop_words="english")
bow = count_vectorizer.fit_transform(texts)
bow_table = pd.DataFrame(
    bow.toarray(),
    index=text_names,
    columns=count_vectorizer.get_feature_names_out()
)
display(bow_table)
```

```python
# Q3(2)
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer(stop_words="english")
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)
tfidf_words = tfidf_vectorizer.get_feature_names_out()
tfidf_table = pd.DataFrame(
    tfidf_matrix.toarray(),
    index=text_names,
    columns=tfidf_words
)
display(tfidf_table.round(3))
```

```python
# Q3(3)
interpretations = [
    "Camera, lens and photos describe the product and its picture quality.",
    "Laptop, processor and coding describe the computer and its main use.",
    "Headphones, cushions and sound describe the product, comfort and audio."
]

for i in range(len(texts)):
    scores = tfidf_matrix[i].toarray()[0]
    top_indices = scores.argsort()[::-1][:3]
    print(text_names[i])
    for index in top_indices:
        print(tfidf_words[index], round(scores[index], 3))
    print(interpretations[i])
    print()

print("TF-IDF gives more weight to words that are frequent in one text and less common across texts.")
```

## Q4. Similarity between two technologies
Write two texts of 4–6 lines each. Preprocess both, calculate Jaccard and TF-IDF cosine similarity, and explain which metric is more useful for these texts.

```python
# Q4(1)
ai_text = """Artificial intelligence helps computers learn patterns from data.
Learning models use data to make predictions.
A model can improve when it receives useful examples.
Developers test a model before using it in an application.
An artificial intelligence system can support people with routine tasks."""

blockchain_text = """Blockchain stores records across a network of computers.
Each block links to earlier records using a digital fingerprint.
The network checks shared records before adding a block.
Developers can build an application that tracks these records.
A blockchain system aims to make shared records difficult to alter."""

print("Artificial intelligence:")
print(ai_text)
print("Blockchain:")
print(blockchain_text)

technology_tokens = []
for text in [ai_text, blockchain_text]:
    tokens = re.findall(r"\b[a-z]+\b", text.lower())
    final_words = []
    for word in tokens:
        if word not in stop_words:
            final_words.append(word)
    technology_tokens.append(final_words)

print("AI tokens:", technology_tokens[0])
print("Blockchain tokens:", technology_tokens[1])
```

```python
# Q4(2)(a)
ai_words = set(technology_tokens[0])
blockchain_words = set(technology_tokens[1])

common_words = ai_words.intersection(blockchain_words)
all_words = ai_words.union(blockchain_words)
jaccard_similarity = len(common_words) / len(all_words)

print("Common words:", sorted(common_words))
print("Jaccard similarity:", round(jaccard_similarity, 4))
```

```python
# Q4(2)(b)
from sklearn.metrics.pairwise import cosine_similarity

clean_technology_texts = []
for tokens in technology_tokens:
    clean_technology_texts.append(" ".join(tokens))

similarity_vectorizer = TfidfVectorizer()
technology_tfidf = similarity_vectorizer.fit_transform(clean_technology_texts)
cosine_score = cosine_similarity(technology_tfidf[0], technology_tfidf[1])[0][0]

print("Cosine similarity:", round(cosine_score, 4))
```

```python
# Q4(2)(c)
print("Jaccard measures the share of distinct words that the texts have in common.")
print("TF-IDF cosine similarity also uses word frequency and reduces the weight of shared words.")
print("Cosine gives better insight here because repeated topic words matter:")
print("AI discusses data and models, while blockchain discusses records and a network.")
print("Both scores measure word overlap; they do not fully understand meaning.")
```

## Q5. Sentiment analysis and a word cloud
Write short product reviews. Use TextBlob to find polarity and subjectivity, classify the reviews, and create a word cloud from the positive reviews.

```python
# Q5(1)
from textblob import TextBlob

reviews = [
    "Excellent study app! The simple design is beautiful, and the helpful reminders make me happy.",
    "The study app is terrible and slow. I hate its confusing design and useless reminders.",
    "The study app has a calendar, a timer, and a notes page."
]

polarities = []
subjectivities = []

for review in reviews:
    sentiment = TextBlob(review).sentiment
    polarities.append(sentiment.polarity)
    subjectivities.append(sentiment.subjectivity)

sentiment_table = pd.DataFrame({
    "Review": reviews,
    "Polarity": polarities,
    "Subjectivity": subjectivities
})
display(sentiment_table)
print("Polarity ranges from -1 for negative to +1 for positive.")
print("Subjectivity ranges from 0 for objective to 1 for subjective.")
```

```python
# Q5(2)
labels = []
positive_reviews = []

for review, polarity in zip(reviews, polarities):
    if polarity > 0:
        labels.append("Positive")
        positive_reviews.append(review)
    elif polarity < 0:
        labels.append("Negative")
    else:
        labels.append("Neutral")

sentiment_table["Sentiment"] = labels
display(sentiment_table)
```

```python
# Q5(3)
from wordcloud import WordCloud
import matplotlib.pyplot as plt

positive_text = " ".join(positive_reviews)
cloud = WordCloud(
    width=800,
    height=400,
    background_color="white",
    colormap="viridis",
    random_state=42
).generate(positive_text)

plt.figure(figsize=(10, 5))
plt.imshow(cloud, interpolation="bilinear")
plt.axis("off")
plt.title("Words from positive reviews")
plt.tight_layout()
plt.show()
```

## Q6. Text generation with a simple Dense model
Use an original paragraph of about 100 words. Tokenize it with Keras Tokenizer, create input sequences, train a simple model, and generate three lines from seed words.

```python
# Q6(1)
from keras.preprocessing.text import Tokenizer

training_text = (
    "Every morning I open my notebook and plan a small project. "
    "I write clear steps before I begin coding. "
    "Each small project helps me learn a useful idea and practice patient thinking. "
    "When a program fails I read the message and check one step at a time. "
    "I ask simple questions and test small changes until the program works. "
    "After a short break I return with fresh attention and improve the code. "
    "I save my work and share what I learned with a friend. "
    "Learning becomes easier when I practice every day and enjoy the small progress I make along the way."
)

tokenizer = Tokenizer()
tokenizer.fit_on_texts([training_text])
token_numbers = tokenizer.texts_to_sequences([training_text])[0]
vocabulary_size = len(tokenizer.word_index) + 1

print("Training paragraph:", training_text)
print("Word count:", len(training_text.split()))
print("Word IDs:", tokenizer.word_index)
print("Tokenized paragraph:", token_numbers)
```

```python
# Q6(2)
import numpy as np
import tensorflow as tf
from keras.preprocessing.sequence import pad_sequences
from keras.models import Sequential
from keras.layers import Embedding, Flatten, Dense

tf.keras.utils.set_random_seed(42)

input_sequences = []
next_words = []

for i in range(1, len(token_numbers)):
    input_sequences.append(token_numbers[max(0, i - 3):i])
    next_words.append(token_numbers[i])

X = pad_sequences(input_sequences, maxlen=3, padding="pre")
y = np.array(next_words)

model = Sequential([
    Embedding(vocabulary_size, 16, input_length=3),
    Flatten(),
    Dense(64, activation="relu"),
    Dense(vocabulary_size, activation="softmax")
])
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("First five input sequences:")
print(X[:5])
print("First five next-word IDs:", y[:5])
model.summary()
```

```python
# Q6(3)
history = model.fit(X, y, epochs=200, batch_size=16, verbose=0)
print("Final training loss:", round(history.history["loss"][-1], 4))
print("Final training accuracy:", round(history.history["accuracy"][-1], 4))

seed_words = ["morning", "learning", "program"]

for seed in seed_words:
    generated_text = seed
    for i in range(12):
        sequence = tokenizer.texts_to_sequences([generated_text])[0]
        sequence = pad_sequences([sequence], maxlen=3, padding="pre", truncating="pre")
        probabilities = model.predict(sequence, verbose=0)[0]
        next_word_id = int(np.argmax(probabilities[1:])) + 1
        next_word = tokenizer.index_word[next_word_id]
        generated_text = generated_text + " " + next_word
    print(generated_text)

print("This small model learns from one paragraph, so it may repeat phrases from its training text.")
```
