```python
import pandas as pd
```

```python
#Q1

fixed_entries = [
    {"question": "what is the annual fee", 
     "answer": "The annual fee is Rs 500.",
     "keywords": "fee cost price charge", 
     "category": "billing"},

    {"question": "how to reset password", 
     "answer": "Go to Settings > Reset Password.",
     "keywords": "password reset login", 
     "category": "account"},

    {"question": "what are your working hours", 
     "answer": "We are open 9 AM to 5 PM.",
     "keywords": "hours timing open time", 
     "category": "general"},

    {"question": "how can i pay the fee", 
     "answer": "You can pay via UPI, card, or net banking.",
     "keywords": "pay payment upi fee", 
     "category": "billing"},
]
```

```python
categories = ["billing", "account", "general"]

roll_number = "1024170396"
L = [int(i) for i in roll_number]
second_last = L[len(L)-2]
last = L[len(L)-1]

cat1 = categories[second_last % 3]
cat2 = categories[last % 3]

print(cat1, cat2)
```

```python
custom_entries = [
    {
        "question": "when is my bill due",
        "answer": "Your bill is due by the 10th of every month.",
        "keywords": "bill due payment",
        "category": cat1
    },
    {
        "question": "how can i view my payment history",
        "answer": "You can view your payment history from the Billing section.",
        "keywords": "payment history billing",
        "category": cat2
    }
]
fixed_entries.extend(custom_entries)
df = pd.DataFrame(fixed_entries)

print(df.to_string())

print(df.shape)
```

```python
#Q2

def query_score(query , df):
    query_set = set(query.lower().split())
    results = []
    number_of_query_words = len(query_set)
    for i in range(len(df)):
        question = df.iloc[i]["question"]
        keywords = df.iloc[i]["keywords"]
        faq_set = set((question + " " + keywords).lower().split())
        matches = query_set & faq_set
        score = len(matches)/number_of_query_words
        if score > 0:
            results.append((score, i))

    results.sort(reverse=True)

    return results

result = query_score("how much is the fee", df)

for score, i in result:
    print("Score:", score)
    print("Question:", df.iloc[i]["question"])
    print("Answer:", df.iloc[i]["answer"])
    print()

```

```python
# Q3
def same_category(category_name, df):
    return df.groupby("category")["question"].get_group(category_name)

result = same_category(cat1, df)
print(result)
```

```python
# Q6

def show_best_matches(query, df):

    results = query_score(query, df)

    if not results:
        print("No matching entries")
        return

    highest_score = results[0][0]

    for score, i in results:
        if score == highest_score:
            print("Question:", df.iloc[i]["question"])
            print("Answer:", df.iloc[i]["answer"])
            print("Score:", score)
            print()

show_best_matches("fee", df)
```
