## Question 3

```python
random.seed(roll_no)

numbers=[random.randint(100,900) for _ in range(100)]
print(numbers)
```

## Question 3 (ii)

```python
odd=[x for x in numbers if x%2]
print(len(odd))
print(odd)
```

## Question 3 (iii)

```python
even=[x for x in numbers if x%2==0]
print(len(even))
print(even)
```

## Question 3 (iv)

```python
def isPrime(n):
    if n<2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True

primes=[x for x in numbers if isPrime(x)]
print(len(primes))
print(primes)
```

## Question 3 (v)

```python
cnt=Counter(numbers)
num,freq=cnt.most_common(1)[0]
print(num)
print(freq)
```
