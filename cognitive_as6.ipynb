```python
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print('Original readings:', temperature)
```

**(a) Add 2°C to every reading using vectorization**

```python
corrected = temperature + 2
print('Corrected readings (°C):', corrected)
```

**(b) Convert Celsius to Fahrenheit:** $F = \frac{9}{5}C + 32$

```python
fahrenheit = (9 / 5) * temperature + 32
print('Readings in °F:', fahrenheit)
```

**(c) Readings greater than 32°C (Boolean indexing)**

```python
mask = temperature > 32
print('Boolean mask:', mask)
print('Readings > 32°C:', temperature[mask])
```

**(d) Count of readings exceeding 32°C**

```python
count = np.sum(temperature > 32)   # True counts as 1
print('Number of readings > 32°C:', count)
```

## Q2. Daily Steps Matrix

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]])
print(steps)
```

```python
print('Total steps:', steps.sum())
```

**(b) Mean number of steps**

```python
print('Mean steps:', steps.mean())
```

**(c) Maximum and minimum values**

```python
print('Maximum:', steps.max())
print('Minimum:', steps.min())
```

**(d) Total steps for each day (`axis=0` — collapses the rows, one total per column/day)**

```python
daily_total = steps.sum(axis=0)
print('Day 1, Day 2, Day 3 totals:', daily_total)
```

**(e) Total steps for each user (`axis=1` — collapses the columns, one total per row/user)**

```python
user_total = steps.sum(axis=1)
print('User 1..4 totals:', user_total)
```

**(f) Position of the maximum value using `argmax()`**

```python
flat_idx = steps.argmax()                       # index in the flattened array
user, day = np.unravel_index(flat_idx, steps.shape)
print('Flat index:', flat_idx)
print(f'Max = {steps[user, day]} at User {user + 1}, Day {day + 1}  (row {user}, col {day})')
```

## Q4.

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]])
y = np.array([40, 65, 30, 85])
```

**(a) Shape and dimensions of X**

```python
print('Shape:', X.shape)
print('Dimensions (ndim):', X.ndim)
```

**(b) Transpose `X.T`**

```python
print(X.T)
```

**(c) Matrix product `X.T @ X`**

```python
XtX = X.T @ X
print(XtX)
```

**(d) Matrix inverse with `np.linalg.inv()`** — valid only when the matrix is square and non-singular, so we check the determinant first.

```python
print('det(XtX) =', np.linalg.det(XtX))
XtX_inv = np.linalg.inv(XtX)
print('Inverse of XtX:\n', XtX_inv)
```

**(e) OLS:** $\beta = (X^T X)^{-1} X^T y$

```python
beta = XtX_inv @ X.T @ y
print('Coefficients (beta):', beta)
print('Sleep coef   :', beta[0])
print('Activity coef:', beta[1])
print('Stress coef  :', beta[2])

# Cross-check with NumPy's built-in least-squares solver
print('lstsq check  :', np.linalg.lstsq(X, y, rcond=None)[0])
```

**(f) Meaning of the coefficients**

Each coefficient is the *change in the predicted assistance score for a one-unit increase in that feature, holding the other two features constant*:

- **Sleep hours coefficient** – effect of one extra hour of sleep on the assistance score.
- **Activity level coefficient** – effect of one extra unit of activity.
- **Stress level coefficient** – effect of one extra unit of stress.

A positive coefficient means the feature *raises* the predicted assistance need; a negative one means it *lowers* it. Larger absolute values mean stronger influence (compare carefully, as the features are on different scales). Note that this model has **no intercept term**, so it forces the prediction to be 0 when all features are 0. With only 4 samples and 3 features the fit is also very small-sample and should be treated as illustrative.

**(g) Prediction for the new user**

```python
new_user = np.array([5, 40, 7])
prediction = new_user @ beta        # same as np.dot(new_user, beta)
print('Predicted assistance score:', prediction)
```
