```python
# Q2
# (a)
arr = np.array([1, 2, 3, 6, 4, 5])
arr = np.flip(arr)
print(arr)

# (b)
def max_frequency(arr):
    values, frequencies = np.unique(arr, return_counts=True)
    max_position = np.argmax(frequencies)
    
    max_value = values[max_position]
    max_frequency = frequencies[max_position]
    indices = np.where(arr == max_value)[0]
    
    return indices, max_value, max_frequency
# i 
x = np.array([1,2,3,4,5,1,2,1,1,1])
print(max_frequency(x))

# ii 
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ]) 
print(max_frequency(y))
```

```python
# Q4
harkamal_singh = np.linspace(10, 100, 25)

print("Array:", harkamal_singh)
print("Dimensions:", harkamal_singh.ndim)
print("Shape:", harkamal_singh.shape)
print("Total elements:", harkamal_singh.size)
print("Data type:", harkamal_singh.dtype)
print("Total bytes:", harkamal_singh.nbytes)

# Reshape the 1-D array into a column array
reshaped_array = harkamal_singh.reshape(25, 1)
print("Reshaped array:")
print(reshaped_array)

# T works as a transpose after reshaping to 2-D
transposed_array = reshaped_array.T
print("Transpose using T:")
print(transposed_array)

# For the original 1-D array, .T does not change its shape
print("Original array using T:", harkamal_singh.T.shape)
```

```python
# Q5
ucs420_harkamal_singh = np.array([[10, 20, 30, 40], [50, 60, 70, 80], [90, 15, 20, 35]])

print("Mean:", np.mean(ucs420_harkamal_singh))
print("Median:", np.median(ucs420_harkamal_singh))
print("Maximum:", np.max(ucs420_harkamal_singh))
print("Minimum:", np.min(ucs420_harkamal_singh))
print("Unique elements:", np.unique(ucs420_harkamal_singh))

reshaped_ucs420_harkamal_singh = ucs420_harkamal_singh.reshape(4, 3)
print("Reshaped array (4 x 3):")
print(reshaped_ucs420_harkamal_singh)

resized_ucs420_harkamal_singh = np.resize(ucs420_harkamal_singh, (2, 3))
print("Resized array (2 x 3):")
print(resized_ucs420_harkamal_singh)
```
